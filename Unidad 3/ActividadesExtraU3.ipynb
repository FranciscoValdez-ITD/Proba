```python
#Ejercicio 1
import math
import numpy as np
import matplotlib.pyplot as plt

# Datos del problema
x = [0, 1, 2, 3, 4, 5]
p = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]

# 1. Verificar suma de probabilidades

print("1) Verificación:")
print("Suma de probabilidades =", sum(p))
print()


# 2. Mostrar distribución

print("2) Distribución de probabilidad:")
for xi, pi in zip(x, p):
    print(f"P(X={xi}) = {pi}")
print()

# 3. Calcular esperanza, varianza y desviación


# Esperanza E(X)
esperanza = sum(x[i] * p[i] for i in range(len(x)))

# E(X^2)
esperanza_cuadrado = sum((x[i]**2) * p[i] for i in range(len(x)))

# Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print("3) Resultados:")
print(f"E(X) = {esperanza:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
print()


# 4. Función acumulada (FDA)

F = np.cumsum(p)

print("4) Función de distribución acumulada:")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# Gráfica FDA
plt.figure()
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Número de células anormales")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()


# 5. Interpretación

print("\n5) Interpretación:")
print(f"- En promedio se esperan {esperanza:.2f} células anormales por muestra.")
print(f"- La variabilidad es {'alta' if desviacion > 1 else 'moderada/baja'} (desviación estándar = {desviacion:.2f}).")
print("- La FDA muestra cómo aumenta la probabilidad acumulada.")
print("- Valores bajos (0,1,2) tienen mayor probabilidad acumulada que valores altos.")
```

```python
# Ejercicio 2
import math
import numpy as np
import matplotlib.pyplot as plt

# Datos del problema
premios = ["A", "B", "C", "D"]
x = [0, 10, 20, 50]  # Montos en pesos
p = [0.50, 0.30, 0.15, 0.05]


# 1. Verificar suma de probabilidades

print("1) Verificación:")
print("Suma de probabilidades =", sum(p))
print()


# 2. Distribución de probabilidad

print("2) Distribución de probabilidad:")
for premio, xi, pi in zip(premios, x, p):
    print(f"P({premio} -> ${xi}) = {pi}")
print()


# 3. Calcular esperanza, varianza y desviación

# Esperanza E(X)
esperanza = sum(x[i] * p[i] for i in range(len(x)))

# E(X^2)
esperanza_cuadrado = sum((x[i]**2) * p[i] for i in range(len(x)))

# Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print("3) Resultados:")
print(f"E(X) = ${esperanza:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
print()


# 4. Función acumulada (FDA)

F = np.cumsum(p)

print("4) Función de distribución acumulada:")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# Gráfica FDA
plt.figure()
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Ganancia (pesos)")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()


# 5. Interpretación

print("\n5) Interpretación:")

print(f"- La ganancia promedio esperada es de ${esperanza:.2f} pesos por juego.")

if esperanza > 0:
    print("- El juego es favorable para el jugador (en promedio gana dinero).")
elif esperanza < 0:
    print("- El juego NO es favorable para el jugador (pierde en promedio).")
else:
    print("- El juego es justo (ni gana ni pierde en promedio).")

print(f"- La variabilidad del juego es de {desviacion:.2f}, lo que indica qué tanto pueden variar los premios.")
print("- La mayoría de las veces se gana poco o nada, ya que las probabilidades más altas están en premios bajos.")
```

```python
#Ejercicio 3
import math
import numpy as np
import matplotlib.pyplot as plt

# Datos del problema
x = [0, 1, 2, 3, 4, 5]  # Unidades vendidas
p = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10]  # Probabilidades

# 1. Verificar suma de probabilidades

print("1) Verificación:")
print("Suma de probabilidades =", sum(p))
print()


# 2. Distribución de probabilidad

print("2) Distribución de probabilidad:")
for xi, pi in zip(x, p):
    print(f"P(X={xi}) = {pi}")
print()


# 3. Calcular esperanza, varianza y desviación


# Esperanza E(X)
esperanza = sum(x[i] * p[i] for i in range(len(x)))

# E(X^2)
esperanza_cuadrado = sum((x[i]**2) * p[i] for i in range(len(x)))

# Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print("3) Resultados:")
print(f"E(X) = {esperanza:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
print()


# 4. Función acumulada (FDA)

F = np.cumsum(p)

print("4) Función de distribución acumulada:")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# Gráfica FDA
plt.figure()
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Unidades vendidas")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()


# 5. Interpretación

print("\n5) Interpretación:")

print(f"- En promedio se espera vender {esperanza:.2f} unidades por semana.")

if desviacion < 1:
    print("- La demanda es bastante estable (baja variabilidad).")
elif desviacion < 2:
    print("- La demanda tiene variabilidad moderada.")
else:
    print("- La demanda es muy variable.")

print(f"- La desviación estándar es {desviacion:.2f}, lo que indica qué tanto varían las ventas.")
print("- La mayor probabilidad está en valores medios (2 y 3 unidades), por lo que esas ventas son las más comunes.")
```
