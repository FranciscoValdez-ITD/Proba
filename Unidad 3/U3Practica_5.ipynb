```python
# Ejercicio 1
import math
import numpy as np
import matplotlib.pyplot as plt

# Datos del problema
x = [0, 1, 2, 3, 4, 5]
p = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]

# 1. Verificar suma de probabilidades
print("1. Verificación:")
print("Suma de probabilidades =", sum(p))
print()

# 2. Mostrar distribución
print("2. Distribución de probabilidad:")
for xi, pi in zip(x, p):
    print(f"P(X={xi}) = {pi}")
print()

# 3. Calcular esperanza, varianza y desviación
esperanza = sum(x[i] * p[i] for i in range(len(x)))
esperanza_cuadrado = sum((x[i]**2) * p[i] for i in range(len(x)))

# Varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)
print("3. Resultados:")
print(f"E(X) = {esperanza:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
print()

# 4. Función acumulada (FDA)
F = np.cumsum(p)
print("4. Función de distribución acumulada:")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# 5. Interpretación
print("5. Interpretación:")
print(f"- En promedio se esperan {esperanza:.2f} células anormales por muestra.")
print(f"- La variabilidad es {'alta' if desviacion > 1 else 'moderada/baja'} (desviación estándar = {desviacion:.2f}).")
print("- La FDA muestra cómo aumenta la probabilidad acumulada.")
print("- Valores bajos (0,1,2) tienen mayor probabilidad acumulada que valores altos.")

# Gráfica FDA
plt.figure()
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Número de células anormales")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos: valores de la variable aleatoria y sus probabilidades
x = np.array([5, 6, 7, 8, 9, 10])
p = np.array([0.05, 0.10, 0.20, 0.30, 0.25, 0.10])

# Cálculo del valor esperado (media)
esperanza = np.sum(x * p)

# Cálculo de la esperanza del cuadrado
esperanza_cuadrado = np.sum(x**2 * p)

# Cálculo de la varianza y desviación estándar
varianza = esperanza_cuadrado - esperanza**2
desviacion = np.sqrt(varianza)

# Opcional: imprimir los resultados para verlos en Colab
print(f"Esperanza (E[X]): {esperanza}")
print(f"Varianza (Var[X]): {varianza}")
print(f"Desviación Estándar (σ): {desviacion}")
```
