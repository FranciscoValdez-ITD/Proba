```python
# 1. Librerias

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr

sns.set(style="whitegrid")
```

```python
# 2. Cargar los archivos y datos iniciales

grupo1 = pd.read_csv('grupo1.csv')
grupo2 = pd.read_csv('grupo2.csv')
grupo3 = pd.read_csv('grupo3.csv')

print("GRUPO 1")
display(grupo1.head())
print("GRUPO 2")
display(grupo2.head())
print("GRUPO 3")
display(grupo3.head())
```

```python
# Ver los datos de todos los grupos

# Grupo 1
print("INFORMACION DEL GRUPO 1")
print(grupo1.info())
print("DESCRIPCION ESTADISTICA")
display(grupo1.describe())

# Grupo 2
print("INFORMACION DEL GRUPO 2")
print(grupo2.info())
print("DESCRIPCION ESTADISTICA")
display(grupo2.describe())

# Grupo 3
print("INFORMACION DEL GRUPO 3")
print(grupo3.info())
print("DESCRIPCION ESTADISTICA")
display(grupo3.describe())
```

```python
# Histogramas

# Grupo 1
plt.figure(figsize=(6,4))
plt.hist(grupo1['Calificacion'], bins=10)
plt.title('Histograma de Calificaciones - Grupo 1')
plt.xlabel('Calificacion')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()

# Grupo 2
plt.figure(figsize=(6,4))
plt.hist(grupo2['Calificacion'], bins=10)
plt.title('Histograma de Calificaciones - Grupo 2')
plt.xlabel('Calificacion')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()

# Grupo 3
plt.figure(figsize=(6,4))
plt.hist(grupo3['Calificacion'], bins=10)
plt.title('Histograma de Calificaciones - Grupo 3')
plt.xlabel('Calificacion')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
# Graficas de dispersion

# Grupo 1
plt.figure(figsize=(6,4))
sns.scatterplot(x=grupo1['HorasEstudio'],y=grupo1['Calificacion'])
plt.title('Diagrama de Dispersion - Grupo 1')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()

# Grupo 2
plt.figure(figsize=(6,4))
sns.scatterplot(x=grupo2['HorasEstudio'],y=grupo2['Calificacion'])
plt.title('Diagrama de Dispersion - Grupo 2')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()

# Grupo 3
plt.figure(figsize=(6,4))
sns.scatterplot(x=grupo3['HorasEstudio'],y=grupo3['Calificacion'])
plt.title('Diagrama de Dispersion - Grupo 3')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()
```

```python
# Regresion Lineal

# Grupo 1 Datos de entrada
horas1 = np.array(grupo1['HorasEstudio']).reshape(-1,1)
calificaciones1 = np.array(grupo1['Calificacion'])

# Crear y entrenar modelo
modelo1 = LinearRegression()
modelo1.fit(horas1, calificaciones1)

# Prediccion para las horas de estudio
prediccion1 = modelo1.predict([[8]])
print("La calificacion esperada para 8 horas de estudio es:",
      prediccion1[0])

# Coeficientes
print("Coeficiente de la pendiente (m):",modelo1.coef_[0])
print("Intercepcion de la recta (b):",modelo1.intercept_)

# Grupo 2
horas2 = np.array(grupo2['HorasEstudio']).reshape(-1,1)
calificaciones2 = np.array(grupo2['Calificacion'])

modelo2 = LinearRegression()
modelo2.fit(horas2, calificaciones2)

prediccion2 = modelo2.predict([[8]])

print("La calificacion esperada para 8 horas de estudio es:",prediccion2[0])
print("Coeficiente de la pendiente (m):",modelo2.coef_[0])
print("Intercepcion de la recta (b):",modelo2.intercept_)

# Grupo 3
horas3 = np.array(grupo3['HorasEstudio']).reshape(-1,1)
calificaciones3 = np.array(grupo3['Calificacion'])

modelo3 = LinearRegression()
modelo3.fit(horas3, calificaciones3)

prediccion3 = modelo3.predict([[8]])

print("La calificacion esperada para 8 horas de estudio es:",prediccion3[0])
print("Coeficiente de la pendiente (m):",modelo3.coef_[0])
print("Intercepcion de la recta (b):",modelo3.intercept_)
```

```python
# Correlacion de Pearson

# Grupo 1
coeficiente1, p_valor1 = pearsonr(grupo1['HorasEstudio'],grupo1['Calificacion'])
print("Coeficiente de correlacion de Pearson:",coeficiente1)
print("Valor p:",p_valor1)

# Grupo 2
coeficiente2, p_valor2 = pearsonr(grupo2['HorasEstudio'],grupo2['Calificacion'])
print("Coeficiente de correlacion de Pearson:",coeficiente2)
print("Valor p:",p_valor2)

# Grupo 3
coeficiente3, p_valor3 = pearsonr(grupo3['HorasEstudio'],grupo3['Calificacion'])
print("Coeficiente de correlacion de Pearson:",coeficiente3)
print("Valor p:",p_valor3)
```

```python
# Graficas con la linea de Regresion

# Grupo 1
plt.figure(figsize=(6,5))
sns.regplot(
    x=grupo1['HorasEstudio'],
    y=grupo1['Calificacion'],
    scatter=True,
    color='blue')

plt.title('Dispersion con Linea de Regresion - Grupo 1')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()

# Grupo 2
plt.figure(figsize=(6,5))
sns.regplot(
    x=grupo2['HorasEstudio'],
    y=grupo2['Calificacion'],
    scatter=True,
    color='blue')

plt.title('Dispersion con Linea de Regresion - Grupo 2')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()

# Grupo 3
plt.figure(figsize=(6,5))
sns.regplot(
    x=grupo3['HorasEstudio'],
    y=grupo3['Calificacion'],
    scatter=True,
    color='blue')

plt.title('Dispersion con Linea de Regresion - Grupo 3')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()
```

**2. ¿Como se interpreta el coeficiente de correlacion?**

El coeficiente de correlacion muestra que tan fuerte es la relacion entre las horas de estudio y las calificaciones, por ejemplo si esta cerca de 1, la relacion es fuerte y positiva, si esta cerca de 0, la relacion es debil y si esta cerca de -1, la relacion es negativa. Entonces si el valor es positivo, significa que al estudiar mas, las calificaciones tambien aumentan.

**3. ¿El modelo de regresion lineal es estadisticamente significativo?**

Si porque si el valor p es menor que 0.05, entoces eso significa que si existe una relacion importante entre las horas de estudio y las calificaciones.

**4. ¿Que tan confiable es el modelo de regresion?**

El modelo es mas confiable cuando la correlacion es alta y los puntos estan cerca de la linea de regresion, esto significa que el modelo puede ayudar a predecir las calificaciones usando las horas de estudio.
